```python
import boto3
import json

# -------------------------------------
# Step 0: AWS Bedrock Client Setup
# -------------------------------------
# Replace with your AWS Access Key and Secret Key
rt = boto3.client(
    "bedrock-runtime",
    region_name="us-east-1",  # Bedrock Claude available in us-east-1 and us-west-2
    aws_access_key_id="AKIAVSX4Z23PH4R4CRWQ",
    aws_secret_access_key="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"
)

# -------------------------------------
# Step 1: List of Products
# -------------------------------------
products = [
    {
        "product_name": "SmartWatch X200",
        "category": "Electronics",
        "features": "Heart rate monitor, GPS, Water-resistant, Long battery life",
        "tone": "Friendly"
    },
    {
        "product_name": "EcoBottle",
        "category": "Kitchen",
        "features": "BPA-free, Reusable, Leak-proof",
        "tone": "Friendly"
    },
    {
        "product_name": "Noise-Canceling Headphones",
        "category": "Electronics",
        "features": "Bluetooth, Long battery life, Comfortable fit",
        "tone": "Professional"
    }
]
```

```python
# -------------------------------------
# Step 2: Few-Shot Examples (Optional)
# -------------------------------------
few_shot_examples = """
Example 1:
Product Name: EcoBottle
Category: Kitchen
Key Features: BPA-free, Reusable, Leak-proof
Tone: Friendly
Description: The EcoBottle is your perfect hydration companion. BPA-free and leak-proof, it keeps your drinks safe and fresh while helping the planet.

Example 2:
Product Name: Noise-Canceling Headphones
Category: Electronics
Key Features: Bluetooth, Long battery life, Comfortable fit
Tone: Professional
Description: Experience crystal-clear sound with our Noise-Canceling Headphones. Enjoy wireless freedom, long-lasting battery, and ultimate comfort during work or travel.
"""
```

```python
# -------------------------------------
# Step 3: Generate Descriptions with Claude 3 Sonnet
# -------------------------------------
for p in products:
    prompt = f"""
You are a creative assistant that writes product descriptions.

{few_shot_examples}

Now create a description for the following:

Product Name: {p['product_name']}
Category: {p['category']}
Key Features: {p['features']}
Tone: {p['tone']}
Description:
""".rstrip()

    # Claude requires 'messages' format
    body = {
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": 300,
        "temperature": 0.7,
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }

    # Invoke Claude 3 Sonnet
    response = rt.invoke_model(
        modelId="anthropic.claude-3-sonnet-20240229-v1:0",
        body=json.dumps(body),
        contentType="application/json"
    )

    # Parse Claude's response
    result = json.loads(response["body"].read())
    description = result["content"][0]["text"]

    print(f"\nProduct: {p['product_name']}")
    print("Generated Description:")
    print(description)
    print("-" * 50)
```
