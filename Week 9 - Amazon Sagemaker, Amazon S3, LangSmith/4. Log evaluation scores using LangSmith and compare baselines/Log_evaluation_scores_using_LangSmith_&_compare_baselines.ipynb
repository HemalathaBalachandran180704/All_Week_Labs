```python
!pip install -U langsmith -q

import os
from langsmith import Client
from datetime import datetime

# Step 1: Set your LangSmith API key (paste yours here)
os.environ["LANGSMITH_API_KEY"] = "ls_api_your_real_key_here"

# Step 2: Initialize client
client = Client()

# Step 3: Create experiment name
experiment_name = f"BERT_LoRA_Comparison_{datetime.now().strftime('%H%M%S')}"

# Step 4: Define evaluation metrics (replace with real results)
bert_metrics = {
    "eval_loss": 0.3927,
    "eval_accuracy": 0.78,
    "model_type": "BERT_Baseline"
}

lora_metrics = {
    "eval_loss": 0.46,
    "eval_accuracy": 0.80,
    "model_type": "BERT_LoRA"
}

# Step 5: Log baseline run
client.create_run(
    name="BERT_Baseline",
    run_type="chain",  # ✅ must be one of: tool, chain, llm, retriever, embedding, prompt, parser
    inputs={"dataset": "Alpaca_Cleaned"},
    outputs={"metrics": bert_metrics},
    project_name=experiment_name,
    metadata={"type": "baseline"}
)

# Step 6: Log LoRA run
client.create_run(
    name="BERT_LoRA",
    run_type="chain",  # ✅ valid run type
    inputs={"dataset": "Alpaca_Cleaned"},
    outputs={"metrics": lora_metrics},
    project_name=experiment_name,
    metadata={"type": "lora"}
)

print(f"✅ Metrics logged to LangSmith project: {experiment_name}")
print("🌐 View here: https://smith.langchain.com/projects")
```
